```python
from dotenv import load_dotenv
load_dotenv(override=True)

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        # temperature/top_p/top_k were removed from messages.create() in
        # anthropic 1.x. claude-sonnet-4-5 still honors them, so pass the
        # value straight through in the request body.
        "extra_body": {"temperature": temperature},
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
# Start with an empty message list
messages = []

# Add the initial user question
add_user_message(messages, "Write an idea for superhero sci-fi movie")

# Low temperature: focused, more repeatable wording
low_temp_answer = chat(messages, temperature=0.0)

print("temperature=0.0:", low_temp_answer)

# High temperature: more varied, more surprising
high_temp_answer = chat(messages, temperature=1.0)

print("temperature=1.0:", high_temp_answer)
```
